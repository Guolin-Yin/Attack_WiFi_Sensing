```python
import pandas as pd
import matplotlib.pyplot as plt
from glob import glob

attack_method = 'fgsm'
smooth_std = 0.05
noise_psr = '*'
noise_std = 1

# Get a list of filenames matching the pattern
filenames = glob(f'*{attack_method}_*{smooth_std}_*{noise_psr}_*{noise_std}_*.xlsx')

def get_noise_psr(filename):
    return float(filename.split('noise_psr_')[1].split('_')[0])

# Sort the filenames by the noise psr value
filenames = sorted(filenames, key=get_noise_psr)

# Create a figure and axis object for the first plot
fig1, ax1 = plt.subplots()

# Create a figure and axis object for the second plot
fig2, ax2 = plt.subplots()

# Iterate over the filenames and plot the data from each file
for filename in filenames:
    # Read the data from the file
    df = pd.read_excel(filename)
    
    # Plot df['acc'][0] - df['acc'] on the first plot
    ax1.plot(df['psr'], df['acc'][0] - df['acc'], label=filename)
    
    # Plot df['acc'] on the second plot
    ax2.plot(df['psr'], df['acc'], label=filename)
    
# Set the title and axis labels for the first plot
ax1.set_title('df[\'acc\'][0] - df[\'acc\']')
ax1.set_xlabel('x')
ax1.set_ylabel('df[\'acc\'][0] - df[\'acc\']')

# Set the title and axis labels for the second plot
ax2.set_title('df[\'acc\']')
ax2.set_xlabel('x')
ax2.set_ylabel('df[\'acc\']')

# Add a legend to the first plot
ax1.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Add a legend to the second plot
ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Show the plots
plt.show()
```

# certified PSR Range

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Reload the data
def draw_certified_psr(dfs,legend):
    plt.figure(figsize=(10, 6))
    colors = ['b', 'g', 'r', 'c', 'm', 'y', 'k']
    
    for i, df in enumerate(dfs):
        # Create a list of PSR thresholds
        psr_thresholds = np.linspace(0, df['PSR'].max(), 100)
        certified_fractions = []

        # Create a function to get fraction of certified samples based on a PSR threshold
        def fraction_certified_psr_threshold(psr_threshold, df):
            # Filter samples that are certified and have PSR above the given threshold
            certified_above_threshold = df[(df["prediction"] != -1) & (df["PSR"] >= psr_threshold)]
            return len(certified_above_threshold) / len(df)

        # Get the fraction of certified samples for a range of PSR thresholds
        psr_values = np.linspace(df["PSR"].min(), df["PSR"].max(), 100)
        fractions_certified = [fraction_certified_psr_threshold(psr, df) for psr in psr_values]

        # Plotting
        plt.plot(psr_values, fractions_certified, marker='o', linestyle='-', color=colors[i%len(colors)], label=legend[i])
        plt.xlabel("PSR Threshold")
        plt.ylabel("Fraction of Samples Certified")
        plt.title("Fraction of Certified Samples by PSR Threshold")
        plt.ticklabel_format(style='sci', axis='x', scilimits=(0,0))
        
        plt.grid(True)
    plt.legend()
    plt.show()
dfs = []
legend = []
for filename in glob(f'../certification_*_psr_*_std_1*276..xlsx'):
    std = float(filename.split('_')[6])
    if  (std > 0.5 and std != 1.5 and std != 1.0) or std == 0.45 or std == 0.35 or std == 0.25:
        continue
    df = pd.read_excel(filename)
    dfs.append(df)
    legend.append(f"std: {filename.split('_')[6]}")
    # legend.append(f"std: {filename.split('_')[6]}, average certified PSR: {df['PSR'].mean():}")
dfs, legend = zip(*sorted(zip(dfs, legend), key=lambda x: float(x[1].split(': ')[1])))
    

draw_certified_psr(dfs,legend)
```
